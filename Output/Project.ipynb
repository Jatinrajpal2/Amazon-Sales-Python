### Extract a Year and  Month No and Name 

```python
df['Year'] = df['Order Date'].dt.year
df['Month No'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()

df
```

## **🔍 Exploratory Data Analysis (EDA)**

### Q1. What are the total Sales, Quantity, and Profit?

```python
print("The Total Sales is:- ", sum(df['Sales'].round(2)))
print("The Total Quantity is:- ", sum(df['Quantity']))
print("The Total Profit is:- ", sum(df['Profit'].round(2)))
```

### Q2. Which are the Top 10 highest selling products?

```python
q2 = df.groupby('Product Name').agg(Highest_Selling_Product = ('Quantity','sum')).sort_values(by='Highest_Selling_Product', 
                                                                                              ascending=False).head(10).reset_index()

q2
```

```python
plt.figure(figsize=(8,5))

a = sns.barplot(data=q2, x='Product Name', y='Highest_Selling_Product',color='Purple' )

for i in a.containers:
    a.bar_label(i)

plt.xticks(rotation= 88)
plt.title("Top 10 Highest Selling Product")
plt.show()
```

### Q3. Which category generates the highest Profit?

```python
q3 = df.groupby('Category').agg(Total_Profit = ('Profit','sum')).sort_values(by='Total_Profit', 
                                                                             ascending=False).round(2).head(1).reset_index()

q3
```

### Q4. State-wise Sales and Profit Analysis

```python
q4 = df.groupby('State').agg(Total_Sales = ('Sales','sum'),
                            Total_Profit = ('Profit','sum')).round(2).reset_index()

q4
```

```python
plt.figure(figsize=(8,4))

sns.lineplot(data=q4, x='State', y='Total_Sales', color='yellow',  
             marker='s', markerfacecolor='blue', label='Total Sales')

sns.lineplot(data=q4, x='State', y='Total_Profit', color='red',  
             marker='s', markerfacecolor='brown', label='Total Profit')

plt.xticks(rotation=75)
plt.legend()
plt.title("Total Sales and Total Profit by State")

plt.show()
```

### Q5. Monthly Sales Trend

```python
q5 = df.groupby(['Month No','Month Name']).agg(Total_Sales = ('Sales','sum')).round(2).sort_values(by='Month No', 
                                                                                                   ascending=True).reset_index()

q5
```

```python
plt.figure(figsize=(8,4))

sns.lineplot(data=q5, x='Month Name', y='Total_Sales', color='red', marker='s', markerfacecolor='blue', label='Total Sales')

plt.xticks(rotation=75)
plt.legend()
plt.title('Monthly Sales Trend')

plt.show()
```

### Q6. Year-wise performance of Sales and Profit

```python
q6 = df.groupby('Year').agg(Total_Sales = ('Sales','sum'),
                           Total_Profit = ('Profit','sum')).round(2).reset_index()


q6
```

```python
plt.figure(figsize=(8,4))

ax = q6.plot(x='Year', y=['Total_Sales','Total_Profit'], kind='bar', width=0.8)


for i in ax.containers:
    ax.bar_label(i)

plt.ylabel("Total Sales and Total Profit")
plt.xticks(rotation=0)
plt.title("Total Sales and Total Profit by Year")

plt.show()
```

## Q7. Which customers placed the highest number of orders?

```python
q7 = df.groupby('EmailID').agg(Highest_Order = ('Order ID','count')).sort_values(by='Highest_Order', 
                                                                                ascending=False).head(1).reset_index()

q7
```

### Q8. Which products have negative profit (loss-making)?

```python
q8 = df.groupby('Product Name').agg(Total_Profit = ('Profit','sum')).round(2)
q8[q8['Total_Profit']<0].reset_index()
```

### Q9. What is the Average Order Value (AOV)?

```python
((df['Sales'].sum()/df['Order ID'].nunique())*1).round(1)
```

```python
q9 = df.groupby('Order ID').agg(Total_Sales = ('Sales', 'sum'))

print("The Average order value is:- ",q9['Total_Sales'].mean().round(1))
```

### Q10. How many repeat customers do we have?

```python
q10 = df.groupby('EmailID').agg(Total_Customer = ('Order ID', 'count'))
q10[q10['Total_Customer']>1].reset_index()
```

### Q11. Which category sells the most quantity?

```python
q11 = df.groupby('Category').agg(Sell_Qty = ('Quantity','sum')).sort_values(by='Sell_Qty', 
                                                                            ascending=False).head(1).reset_index()


q11
```

### Q12. Calculate Profit Margin % for each product

```python
q12 = df.groupby('Product Name').agg(Total_Sales = ('Sales','sum'),
                                     Total_Profit = ('Profit', 'sum')).round(2).reset_index()

q12['Profit Margin%'] = ((q12['Total_Profit']/q12['Total_Sales'])*100).round(2)

q12
```

### Q13. Which is the most profitable month?

```python
q13 = df.groupby('Month Name').agg(Total_Profit = ('Profit', 'sum')).sort_values(by='Total_Profit', 
                                                                                 ascending=False).head(1).round(2).reset_index()

q13
```

### Q14. What is the average shipping time?

```python
df['Ship Time'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
print("The Average Shipping Time:- ",df['Ship Time'].mean().round(2))
```

### Q15. Does Shipping Time impact Profit? (Correlation)

```python
df[['Ship Time','Profit']].corr()
```
